### 🟡 E2 — Cuidado: todo viene como texto
Sumá la columna `puntaje` de todas las películas. Vas a chocar con que los números vienen como **texto**. Resolvelo.

**Listo cuando:** la suma da un número y entendés por qué había que convertir.

```python
import csv

with open('peliculas.csv', 'r', encoding='utf-8') as file:
    reader = csv.reader(file)
    next(reader)
    total = []
    for row in reader:
        total.append(float(row[2]))

print(sum(total))

```

### 🔴 E3 — Reporte
Imprimí: cuántas películas hay, el puntaje promedio, y el título de la mejor puntuada.

**Listo cuando:** los tres datos salen correctos del archivo real.

```python
import csv

with open('peliculas.csv', 'r', encoding='utf-8') as file:
    reader = csv.reader(file)
    next(reader)
    pelis = list(reader)
def promedio(lista):
    total = []
    for row in lista:
        total.append(float(row[2]))
    return sum(total)/len(total)
def mejor(lista):
    mej = 0.0
    til = ""
    for i in lista:
        if float(i[2]) >= mej:
            mej = float(i[2])
            til = i[0]
    return til
def contador(lista):
    return len(lista)
def reporte(lista):
    
    return f"total: {contador(lista)}, puntaje promedio: {promedio(lista)}, mejor titulo: {mejor(lista)}"

print(reporte(pelis))
```

### 🟡 E4 — Filtrar y guardar
Agregale una columna `genero` **al mismo** `peliculas.csv` que ya venís usando desde E1 (no crees un archivo nuevo con otro nombre — si lo hacés, terminás con dos versiones del dataset y los resultados de E1-E3 y E4-E5 dejan de coincidir). Editalo a mano. Leé el archivo, quedate solo con las películas de un género elegido, y escribí un **nuevo CSV** (`filtradas.csv`) solo con esas.

**Listo cuando:** el nuevo CSV existe, tiene menos filas que el original (que ahora incluye la columna `genero`), y lo pueden abrir y verificar a ojo.

```python
import csv

with open('peliculas.csv', 'r') as field:
    reader = csv.reader(field)
    encabezado = next(reader)
    todas_las_filas = []

    for fila in reader:
        todas_las_filas.append(fila)

with open('filtradas.csv', mode='w', newline='') as archivo_csv:
    escritor_csv = csv.writer(archivo_csv)
    escritor_csv.writerow(encabezado)  

    for i in todas_las_filas:
        if i[-1] == 'accion':
            escritor_csv.writerow(i)
```

### 🔴 E5 — Agrupar por categoría
Usando el CSV con `genero` de E4, calculá el **puntaje promedio por género** (un diccionario `{genero: promedio}`).

**Listo cuando:** cada género del archivo aparece una vez en el resultado, con su promedio correcto.

```python
import csv

suma_por_genero = {}
cantidad_por_genero = {}

with open('peliculas.csv', 'r', encoding='utf-8') as file:
    reader = csv.DictReader(file)
    for row in reader:
        genero = row['genero']
        puntaje = float(row['puntaje'])
        
        # si el género no existe todavía, .get() devuelve 0 y arranca desde ahí
        suma_por_genero[genero] = suma_por_genero.get(genero, 0) + puntaje
        cantidad_por_genero[genero] = cantidad_por_genero.get(genero, 0) + 1

# ahora divido suma/cantidad para cada género
promedio_por_genero = {}
for genero in suma_por_genero:
    promedio_por_genero[genero] = suma_por_genero[genero] / cantidad_por_genero[genero]

print(promedio_por_genero)
```
